## 1. Parametros de entrada

```python
dbutils.widgets.text("ambiente", "dev")
var_ambiente = dbutils.widgets.get("ambiente")
```

## 2. Constantes y variables

```python
TBL_PAGO_FIN = f"mb_gold_{var_ambiente}.inof.td_pago_servicio"
```

## 3. Creacion de la tabla

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {TBL_PAGO_FIN} (
    Cod_Pago_Servicio        STRING         COMMENT 'Codigo del pago de servicio',
    fec_pago_servicio        TIMESTAMP      COMMENT 'Fecha del pago',
    nom_empresa_servicio     STRING         COMMENT 'Nombre de la empresa',
    mto_servicio_origen      DOUBLE         COMMENT 'Monto del servicio en moneda origen',
    nro_documento_cliente    STRING         COMMENT 'Numero de documento del cliente',
    _ingestion_time          TIMESTAMP      COMMENT 'Marca de ingesta',
    _processing_time         TIMESTAMP      COMMENT 'Marca de procesamiento'
)
USING DELTA
COMMENT 'Tabla de pagos de servicio'
TBLPROPERTIES (
    'frecuencia' = 'diaria',
    'naturaleza' = 'transaccional',
    'tipo_tabla' = 'hechos',
    'owner'      = 'inof',
    'dac'        = 'si'
)
""")
```
